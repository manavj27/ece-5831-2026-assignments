Setup
Import NumPy and the LogicGate class, and list the four possible binary inputs.

```python
import numpy as np
from logic_gate import LogicGate

gate = LogicGate()
inputs = [(0, 0), (0, 1), (1, 0), (1, 1)]
print(np.__version__)
```

Perceptron with NumPy

Instead of writing x1*w1 + x2*w2, the inputs and weights are stored as NumPy arrays.
w * x multiplies elementwise, and np.sum(w * x) adds the products, which gives the same result
as the dot product np.dot(w, x). The bias b is then added, and the output is 1 if the result is
greater than 0.

```python
x = np.array([1, 1])
w = np.array([0.5, 0.5])
b = -0.7
print("w * x        =", w * x)
print("np.sum(w*x)  =", np.sum(w * x))
print("np.dot(w, x) =", np.dot(w, x))
print("sum + b      =", np.sum(w * x) + b)
```

AND Gate

Weights w = [0.5, 0.5], bias b = -0.7.

Only the input (1, 1) gives 0.5 + 0.5 - 0.7 = 0.3 > 0. A single active input gives
0.5 - 0.7 = -0.2, so the output is 1 only when both inputs are 1.

```python
for x1, x2 in inputs:
    print(x1, x2, "->", gate.and_gate(x1, x2))
```

NAND Gate

Weights w = [-0.5, -0.5], bias b = 0.7.

NAND is NOT AND, so the weights and bias of the AND gate are negated. The output is 0 only when both
inputs are 1 (-1.0 + 0.7 = -0.3).

```python
for x1, x2 in inputs:
    print(x1, x2, "->", gate.nand_gate(x1, x2))
```

OR Gate

Weights w = [0.5, 0.5], bias b = -0.2.

A single active input gives 0.5 - 0.2 = 0.3 > 0, so the output is 1 whenever at least one input is 1.

```python
for x1, x2 in inputs:
    print(x1, x2, "->", gate.or_gate(x1, x2))
```

NOR Gate

Weights w = [-0.5, -0.5], bias b = 0.2.

NOR is NOT OR, so the weights and bias of the OR gate are negated. The output is 1 only when both
inputs are 0.

```python
for x1, x2 in inputs:
    print(x1, x2, "->", gate.nor_gate(x1, x2))
```

XOR Gate
A single perceptron is a linear classifier: it splits the input space with one straight line. XOR's
outputs (1 for (0, 1) and (1, 0), 0 for (0, 0) and (1, 1)) cannot be separated by one line,
so no single choice of 'w' and 'b' works.

The solution is a two-layer perceptron:

- Layer 1: s1 = NAND(x1, x2) and s2 = OR(x1, x2)
- Layer 2: y = AND(s1, s2)

The table below shows the hidden-layer outputs s1 and s2 along with the final output y.

```python
print("x1 x2 | s1 s2 | y")
for x1, x2 in inputs:
    s1 = gate.nand_gate(x1, x2)
    s2 = gate.or_gate(x1, x2)
    print(f" {x1}  {x2} |  {s1}  {s2} | {gate.xor_gate(x1, x2)}")
```
