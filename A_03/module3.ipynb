```python
# ECE 5831 - Assignment 3: Perceptron Logic Gates
# The LogicGate class (logic_gate.py) builds logic gates from perceptrons using NumPy.
# A perceptron outputs y = 1 if sum(w * x) + b > 0, otherwise y = 0.

import numpy as np
from logic_gate import LogicGate

gate = LogicGate()
inputs = [(0, 0), (0, 1), (1, 0), (1, 1)]
print(np.__version__)
```

```python
# Perceptron with NumPy
# Instead of x1*w1 + x2*w2, the inputs and weights are NumPy arrays.
# np.sum(w * x) does elementwise multiplication and then sums (same as np.dot(w, x)).
x = np.array([1, 1])
w = np.array([0.5, 0.5])
b = -0.7
print("w * x        =", w * x)
print("np.sum(w*x)  =", np.sum(w * x))
print("np.dot(w, x) =", np.dot(w, x))
print("sum + b      =", np.sum(w * x) + b)
```

```python
# AND gate: w = [0.5, 0.5], b = -0.7
# Only (1, 1) gives 0.5 + 0.5 - 0.7 = 0.3 > 0, so only (1, 1) outputs 1.
for x1, x2 in inputs:
    print(x1, x2, "->", gate.and_gate(x1, x2))
```

```python
# NAND gate: w = [-0.5, -0.5], b = 0.7
# NAND is NOT AND, so the weights and bias of AND are negated.
for x1, x2 in inputs:
    print(x1, x2, "->", gate.nand_gate(x1, x2))
```

```python
# OR gate: w = [0.5, 0.5], b = -0.2
# One active input gives 0.5 - 0.2 = 0.3 > 0, so any input of 1 outputs 1.
for x1, x2 in inputs:
    print(x1, x2, "->", gate.or_gate(x1, x2))
```

```python
# NOR gate: w = [-0.5, -0.5], b = 0.2
# NOR is NOT OR, so the weights and bias of OR are negated.
for x1, x2 in inputs:
    print(x1, x2, "->", gate.nor_gate(x1, x2))
```

```python
# XOR gate: a single perceptron cannot do XOR because the outputs are not linearly separable.
# A two-layer perceptron solves it: s1 = NAND(x1, x2), s2 = OR(x1, x2), y = AND(s1, s2).
print("x1 x2 | s1 s2 | y")
for x1, x2 in inputs:
    s1 = gate.nand_gate(x1, x2)
    s2 = gate.or_gate(x1, x2)
    print(f" {x1}  {x2} |  {s1}  {s2} | {gate.xor_gate(x1, x2)}")
```
