```python
import numpy as np
from multilayer_perceptron import MultilayerPerceptron

mlp = MultilayerPerceptron()
print(np.__version__)
```

```python
for key, value in mlp.net.items():
    print(f"{key}: shape {value.shape}")
    print(value)
    print()
```

```python
x = np.array([1.0, 0.5, -0.5])

a1 = np.dot(x, mlp.net['w1']) + mlp.net['b1']
z1 = mlp.sigmoid(a1)
print("Layer 1 (sigmoid):", z1)

a2 = np.dot(z1, mlp.net['w2']) + mlp.net['b2']
z2 = mlp.sigmoid(a2)
print("Layer 2 (sigmoid):", z2)

a3 = np.dot(z2, mlp.net['w3']) + mlp.net['b3']
z3 = mlp.sigmoid(a3)
print("Layer 3 (sigmoid):", z3)

a4 = np.dot(z3, mlp.net['w4']) + mlp.net['b4']
y = mlp.relu(a4)
print("Layer 4 (relu):   ", y)
```

```python
y_forward = mlp.forward(x)
print("forward(x):", y_forward)
print("Matches step-by-step result:", np.allclose(y, y_forward))
```

```python
inputs = [
    np.array([1.0, 0.5, -0.5]),
    np.array([0.0, 0.0, 0.0]),
    np.array([1.0, 1.0, 1.0]),
    np.array([-1.0, 2.0, 0.3]),
]
for x in inputs:
    print(f"x = {x} -> y = {mlp.forward(x)}")
```

```python
batch = np.array(inputs)
y_batch = mlp.forward(batch)
print("Batch input shape:", batch.shape)
print("Batch output shape:", y_batch.shape)
print(y_batch)
```
